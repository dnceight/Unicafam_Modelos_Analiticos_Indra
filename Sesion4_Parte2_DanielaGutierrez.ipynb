CLASIFICADOR DE NOTAS

```python
#Crea una función que reciba una nota de 0 a 5 y devuelva el estado del estudiante. Luego aplícala a una lista de notas.

#  Nota < 3.0 → "Reprueba"; de 3.0 a 4.4 → "Aprueba"; ≥ 4.5 → "Excelente".
#  Valida: si la nota no está entre 0 y 5, devuelve "Nota inválida".
#  Aplica la función con un for a [2.5, 3.8, 4.7, 5.2, 3.0].
#  Cuenta cuántos estudiantes aprobaron.
```

```python
def estado_estudiante(nota):
  if nota < 0 or nota > 5:
    return "Nota inválida"
  elif nota < 3:
    return "Reprueba"
  elif nota <= 4.4:
    return "Aprueba"
  else:
    return "Excelente"
```

```python
notas_curso=[2.5, 3.8, 4.7, 5.2, 3.0]
Aprobados = 0

for n in notas_curso:
  print(f"Con la nota {n} el estado del estudiante es: {estado_estudiante(n)}")
  if estado_estudiante(n) == "Aprueba" or estado_estudiante(n) == "Excelente":
    Aprobados += 1

print("Aprobados: ", Aprobados)
```

CARRITO DE COMPRAS

```python
#Una tienda online guarda el carrito como un diccionario producto → (precio, cantidad). Calcula el total de la compra.

#  Carrito: camiseta (45.000 × 2), jean (120.000 × 1), medias (12.000 × 3).
#  Recorre el diccionario y calcula el subtotal.
#  Si el subtotal supera 200.000, aplica 10% de descuento.
#  Suma IVA del 19% e imprime el total con formato de pesos.
```

```python
carrito = {
    "camiseta": {45000, 2},
    "jean": { 120000,  1},
    "medias": { 12000, 3},
}
```

```python
subtotal = 0
for producto, (precio, cantidad) in carrito.items():
    subtotal += precio * cantidad
```

```python
descuento = 0
if subtotal > 200000:
    descuento = subtotal * 0.10

subtotal_con_descuento = subtotal - descuento
```

```python
iva = subtotal_con_descuento * 0.19
total = subtotal_con_descuento + iva
```

```python
def formato_pesos(valor):
    return f"${valor:,.0f}".replace(",", ".")

print(f"Subtotal:  {formato_pesos(subtotal)}")
print(f"Descuento: {formato_pesos(descuento)}")
print(f"IVA (19%): {formato_pesos(iva)}")
print(f"Total:     {formato_pesos(total)}")
```

NOMINA CON PANDAS

```python
# Con el DataFrame de empleados de la actividad, responde preguntas de recursos humanos.

#  Agrega 3 empleados nuevos usando pd.concat o creando la tabla completa.
#  Crea la columna bono: 10% del salario si el área es Ventas, 5% si no.
#  Calcula el costo total de nómina mensual por área.
#  ¿Quién es el empleado con mayor salario?
```

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame({
    "nombre": ["Ana", "Luis", "Marta", "Juan", "Sofía"],
    "area": ["Ventas", "TI", "Ventas", "TI", "RR. HH."],
    "salario": [2800, 4500, 3100, 5200, 3000],
})
```

```python
# 1. Agregar 3 empleados nuevos usando pd.concat
nuevos_empleados = pd.DataFrame({
    'nombre': ['Pedro', 'Laura', 'Jorge'],
    'area': ['Ventas', 'Finanzas', 'TI'],
    'salario': [3200, 3800, 4100],
})

df = pd.concat([df, nuevos_empleados], ignore_index=True)
```

```python
# 2. Crear la columna bono: 10% del salario si el área es Ventas, 5% si no
df['bono'] = np.where(
    df['area'] == 'Ventas', df['salario'] * 0.10, df['salario'] * 0.05
)
```

```python
# 3. Calcular el costo total de nómina mensual por área (salario + bono)
df['total_mensual'] = df['salario'] + df['bono']
nomina_por_area = df.groupby('area')['total_mensual'].sum().reset_index()
```

```python
# 4. ¿Quién es el empleado con mayor salario?
empleado_top = df.loc[df['salario'].idxmax()]

print('--- DATAFRAME ACTUALIZADO ---')
print(df)
print('\n--- NÓMINA MENSUAL POR ÁREA ---')
print(nomina_por_area)
print(
    f"\n--- EMPLEADO CON MAYOR SALARIO ---\n{empleado_top['nombre']}"
    f" (${empleado_top['salario']})"
)
```
