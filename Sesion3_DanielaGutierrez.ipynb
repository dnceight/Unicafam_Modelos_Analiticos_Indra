¿TE ALCANZA PARA EL CAFE?

```python
presupuesto = 60000
precio_cafe = 2100
cafes_semana = 5

#¿Cuánto gastas al mes? (4 Semanas)
gasto_mes = precio_cafe * cafes_semana * 4

#¿Cuanto te sobra?
sobra = presupuesto - gasto_mes

#¿Te alcanza?
alcanza = presupuesto >= gasto_mes

print(gasto_mes, sobra, alcanza)

```
